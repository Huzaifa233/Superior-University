```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
```

```python
nltk.download("punkt")
nltk.download("Stopwords")
nltk.download("wordnet")
nltk.download("average_perceptron_tagger")

print("Setup complete")
```

### Input Text:

```python
text  = "The cats were running faster than the dogs, but one of the dogs was sleeping peacefully."

print("Original Text:\n", text)
```

### Tokenization:

```python
tokens = word_tokenize(text)
print("Token:")
print(tokens)
```

### Stopword Removal:

```python
stopwords = set(stopwords.words("english"))
filter_tokens = [word for word in tokens if word.lower() not in stopwords]
print("Filtered Tokens: ")
print(filter_tokens)
```

### Stemming:


```python
ps = PorterStemmer()

stemmed_tokens = [ps.stem(word) for word in filter_tokens]

print("Stemmed Tokens:")
print(stemmed_tokens)
```

### Lemmatization:

```python
lemmatizer = WordNetLemmatizer()

lemmatized_tokens = [lemmatizer.lemmatize(word) for word in filter_tokens]

print("Lemmatized Tokens:")
print(lemmatized_tokens)
```

### POS Tagging:

```python
pos_tags = nltk.pos_tag(tokens)

print("POS Tags:")
print(pos_tags)
```
