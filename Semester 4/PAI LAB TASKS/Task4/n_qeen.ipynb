```python
def print_board(board, n):
    for i in range(n):
        for j in range(n):
            if board[i][j] == 1:
                print(" Q ", end="")
            else:
                print("   ", end="")

            if j < n-1:
                print("|", end="")
        print()

        if i < n-1:
            print("---" * n  + "-"*(n-1))
```

```python
def is_safe(board, row, col, n):
    for i in range(row):
        if board[i][col] == 1:
            return False
        
    i, j = row, col
    while i >=0 and j>=0:
        if board[i][j] == 1:
            return False
        i -=1
        j -=1

    i, j = row, col
    while i >=0 and j<n:
        if board[i][j] == 1:
            return False
        i -=1
        j +=1
    return True
```

```python
def Solve_n_Queens(board, row, n):
    if row >=n:
        print_board(board, n)
        print("\n" + "="*20 + "\n")
        return True

    res = False
    for col in range(n):
        if is_safe(board, row, col, n):
            board[row][col] = 1
            res = Solve_n_Queens(board, row+1, n) or res
            board[row][col] = 0
    return res
```

```python
n = 4
board = [[0]*n for _ in range(n)]
if not Solve_n_Queens(board, 0, n):
    print("NO SOLUTION EXISTS..")
```
