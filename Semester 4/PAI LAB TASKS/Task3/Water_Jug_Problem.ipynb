```python
# Jug Capacity 
capacity_A = 4
capacity_B = 3
goal = 2

exp_states = set()
```

```python
    
def generate_next_states(curr):
    A, B = curr
    opt = []

    opt.append(((capacity_A, B), "Fill Jug A"))
    opt.append(((A, capacity_B), "Fill Jug B"))
    opt.append(((0, B), "Empty Jug A"))
    opt.append(((A, 0), "Empty Jug B"))

    pour_A_to_B = min(A, capacity_B - B)
    opt.append(((A - pour_A_to_B, B + pour_A_to_B), "Pour from A to B"))

    pour_B_to_A = min(B, capacity_A - A)
    opt.append(((A + pour_B_to_A, B - pour_B_to_A), "Pour from B to A"))

    return opt


```

```python
def dfs_search(state, steps):
    if state in exp_states:
        return False
    exp_states.add(state)

    A, B = state

    if A == goal or B == goal:
        for s in steps:
            print(s)
        print(f"Goal Achieved: {state}")
        return True

    for next_state, action in generate_next_states(state):
        if dfs_search(next_state, steps + [f"{action} -> {next_state}"]):
            return True

    return False
```

```python
start_state = (0, 0)
dfs_search(start_state, [])
```
